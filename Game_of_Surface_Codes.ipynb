```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# Qubits 0 and 1 hold data; qubit 2 is the ancilla.
qc = QuantumCircuit(3, 1)

# qc.x(0)          # Prepare data as q0=1, q1=0.
qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)      # Accumulate both data bits into the ancilla.
qc.cx(1, 2)
qc.measure(2, 0)

backend = AerSimulator()
result = backend.run(
    transpile(qc, backend), shots=100
).result()

print(qc.draw())
print(result.get_counts())
```
